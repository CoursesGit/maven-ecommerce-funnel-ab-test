# Business Recommendation

本 Notebook 汇总前序数据审计、实验有效性检查与指标分析的已有结果，不重新计算实验指标。

## Business Question

历史 Billing Page 实验是否为新版页面在当前流量范围内的逐步推广提供了证据？

## Analysis Objective

将实验结果转化为审慎、可执行的业务建议，并明确证据边界与后续监控要求。

## Data Used

仅引用 `04_Experiment_Analysis.ipynb` 已输出的 Primary、Diagnostic、Guardrail、Bootstrap 与 Sensitivity Analysis 结果。

## Key Metrics

- Primary: Billing-to-Purchase Conversion Rate
- Guardrails: Revenue per Billing Session、Net Revenue per Billing Session、Refund Rate


## 1. Executive Summary

本项目基于 Maven Fuzzy Factory 电商网站的历史 Billing Page 实验数据，对 `/billing`（Control）与 `/billing-2`（Treatment）的表现进行复盘。实验对象为在实验窗口内首次进入任一 Billing 页面的网站 Session；实际曝光时间覆盖 **2012-09-10 00:13:05 至 2012-11-09 22:11:29**。

有效性检查未发现 SRM 或已检查协变量的明显失衡。主指标结果显示 Treatment 的 Billing-to-Purchase Conversion Rate 高于 Control，差异具有统计显著性；三套跨组用户敏感性口径的方向和显著性一致。收入护栏指标同步为正，但 Treatment 的退款率更高，需要在扩大流量时重点监控。


## 2. Experiment Result Summary

### Primary Metric

| Metric | Control | Treatment | Difference |
|---|---:|---:|---:|
| Billing Sessions | 657 | 654 | - |
| Purchases | 300 | 410 | - |
| Billing-to-Purchase Conversion Rate | 45.66% | 62.69% | **+17.03 pp** |
| Relative Lift | - | - | **+37.29%** |

两比例 Z 检验：**z = 6.1872，p = 6.12e-10**。Treatment 相对 Control 的绝对转化率差异 95% CI 为 **[+11.71 pp, +22.34 pp]**。

### Guardrail Metrics

| Metric | Control | Treatment | Observed Difference |
|---|---:|---:|---:|
| Revenue per Billing Session | $22.83 | $31.34 | +$8.51 |
| Net Revenue per Billing Session | $21.30 | $28.36 | +$7.05 |
| Refund Rate | 6.67% | 9.51% | +2.85 pp |

已有 Bootstrap 结果显示，Revenue per Billing Session 差异的 95% CI 为 **[$5.85, $11.13]**，Net Revenue per Billing Session 差异的 95% CI 为 **[$4.37, $9.69]**。Treatment退款率比Control高**2.85个百分点**；以购买订单为分母的双侧两比例Z检验得到 **z = 1.3568，p = 0.1748**，差异95% CI为 **[-1.16 pp, +6.85 pp]**。差异未达到5%显著性水平，但这不能证明退款风险不存在，推广过程中仍需持续监控。


## 3. Business Interpretation

当前证据支持新版 Billing 页面在本次历史实验覆盖的流量范围内继续扩大验证，而不是直接全面上线。收益主要体现为：更多到达 Billing 页面的 Session 完成购买，并进一步反映在每个 Billing Session 的毛收入和净收入提升上。

这些结果只能说明两个页面版本与结果指标之间的实验表现差异，不能据此推断具体页面元素、文案、布局或交互机制是提升原因。Treatment 退款率较高也是推广决策中的重要风险信号，需要随流量扩大持续观察。


## 4. Limitations

- 本分析是基于历史实验数据的复盘，并非事前设计或实时执行实验。
- 缺少原始随机化日志和事前实验方案，无法完全验证分流机制与所有实施细节。
- 实验单位为 Session；同一用户可能贡献多个 Session，因此不能把结果直接解释为独立用户层面的效果。
- 数据中存在少量用户跨组曝光。敏感性分析结果方向一致，但跨组情况仍可能带来污染。
- 当前窗口只能支持本次实验期内的短期购买与收入表现，无法证明长期复购、客户价值或长期退款效果。
- Maven Fuzzy Factory为教学模拟数据，当前观察到的效果幅度可能高于真实线上环境，不应直接作为其他业务场景的提升预期。


## 5. Recommendation

1. 建议采用分阶段方式逐步扩大新版 Billing 页面的流量，并保留稳定对照流量以持续验证效果。
2. 推广期间同时监控 Billing-to-Purchase Conversion Rate、Revenue per Billing Session、Net Revenue per Billing Session 和 Refund Rate，避免只优化转化率。
3. 为关键指标设置预警与回退条件，特别关注退款率是否随流量扩大继续上升。
4. 后续实验应保留随机化与曝光日志，明确事前指标、样本量和停止规则，并减少用户跨组污染。
5. 在更长观察窗口内验证复购、长期净收入和退款表现，再决定是否进一步扩大覆盖范围。
