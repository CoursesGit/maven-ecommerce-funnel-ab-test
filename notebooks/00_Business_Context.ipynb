# Business Context & Experiment Objective

## Business Background

Maven Fuzzy Factory 是一个模拟电商网站。本项目基于网站已有的历史 Billing Page 实验曝光，复盘 `/billing` 与 `/billing-2` 两个版本的购买转化和商业表现。项目不声称设计实验、执行随机分流或负责实验上线。

## Analysis Objective

- 说明 Billing 环节在购买路径中的业务位置。
- 从历史页面曝光重构 Session 级实验样本。
- 在计算效果前检查 SRM、协变量平衡与跨组曝光。
- 评估转化、收入及退款护栏指标。
- 基于证据提出逐步推广与后续验证建议。

## Experiment Overview

| Item | Definition |
|---|---|
| Control | `/billing` |
| Treatment | `/billing-2` |
| Experiment unit | Session-level Billing Page exposure |
| Group reconstruction | Session 首次符合条件的 Billing 页面曝光 |
| Analysis window | 2012-09-10（含）至 2012-11-10（不含） |

由于没有原始随机化日志和事前设计文档，实验分组只能通过历史曝光重构；该限制贯穿后续有效性检查与结论解释。

## Analysis Framework

1. **Business Context**：明确业务问题、实验对象和证据边界。
2. **Data Understanding**：审计四张核心表、字段、关系和质量。
3. **Funnel Diagnosis**：说明 Billing 页面为何值得评估，不将漏斗结果视为实验结果。
4. **Experiment Reconstruction**：以首次 Billing 曝光构造一行一个 Session 的实验底表。
5. **Experiment Validity Check**：检查 SRM、协变量平衡、Session 跨版本和用户跨组。
6. **Metric Analysis**：分析 Primary、Diagnostic 和 Guardrail 指标及敏感性结果。
7. **Business Recommendation**：在限制条件下提出分阶段推广建议。
