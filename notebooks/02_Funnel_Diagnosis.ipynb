# Funnel Diagnosis

## Objective

使用实验前的固定流量 cohort 描述购买路径流失，说明 Billing 页面为什么具有实验评估价值。该模块是业务诊断背景，不是 Billing Page 实验本身，也不用于证明 Treatment 的效果。

## Cohort and Method

- 时间窗口：2012-08-05（含）至 2012-09-05（不含）
- 流量范围：`gsearch` + `nonbrand`
- 分析单位：Session
- 页面顺序依据：`website_pageview_id`
- 同一 Session 在每个阶段最多计数一次

计算逻辑位于 `../sql/02_funnel_diagnosis.sql`，现有结果保存在 `../output/tables/funnel_metrics.csv`。

## Purchase Funnel

```text
Session / Landing
        ↓
Product
        ↓
Cart
        ↓
Shipping
        ↓
Billing
        ↓
Purchase
```

项目 SQL 进一步保留 Product Detail 阶段，以确保页面到达顺序完整。各阶段定义如下：

| Stage | Page definition |
|---|---|
| Session / Landing | `/lander-1` |
| Product | `/products` and `/the-original-mr-fuzzy` |
| Cart | `/cart` |
| Shipping | `/shipping` |
| Billing | `/billing` |
| Purchase | `/thank-you-for-your-order` |

## Why Billing Is Worth Evaluating

Billing 位于购买路径末端，到达该页面的 Session 已表现出较强购买意图；从 Billing 到 Purchase 仍存在明显流失，因此这一环节具有直接商业价值。同时，历史数据提供 `/billing` 与 `/billing-2` 两种曝光，使后续能够重构页面版本比较。

这一诊断只回答“为什么值得评估”。它不能说明新版页面有效，也不能替代实验样本重构、有效性检查或统计检验。
